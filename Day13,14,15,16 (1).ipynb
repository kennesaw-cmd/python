# Exploaratory Data Analysis (EDA)

- Univariate Analysis (Single Variable Analysis)
- Bivariate Analysis (Two Variable Analysis)
- Multivariate Analysis (Multiple Variable Analysis)

```python
df = pd.read_csv('Cleaned_pizza_sales.csv')
df.sample()
```

## 1. Univariate Analysis

```python
# Seaborn over Matplotlib : Seaborn is Pretty version of Matplotlib

# Seaborn is a higher level interface of Matplotlib
```

- Countplot (For categorical columns) | pizza_size

```python
# Displays the count of each pizza size

import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='pizza_size', data=df, palette='Blues')
plt.title('Frequency of Pizza Sizes')
plt.xlabel('Pizza Size')
plt.ylabel('No of Pizza sold')
```

- Countplot (For categorical columns) | pizza_category

```python
sns.countplot(x='pizza_category', data=df, palette='Set1')
# plt.legend(title='Subcategories')
plt.title('Frequency of Pizza Categories')
plt.xlabel('Pizza Categories')
plt.ylabel('No of Pizza sold.')
```

- Pie Chart (For categorical columns) | Pizza Categories

```python
plt.figure(figsize=(6,6))
size_count = df['pizza_category'].value_counts()

plt.pie(size_count.values, labels=size_count.index, autopct='%1.0f%%')
plt.title('Proportion of each pizza category')
```

- Histogram (For continuous column) | Unit Price

```python
sns.histplot(x='unit_price', data=df, bins=10)
plt.title('Dist of Unit Prices of Pizza')
plt.xlabel('Unit Prices')
```

- - Violin Plot (For continuous column) | Unit Price

```python
sns.violinplot(y='unit_price', data=df)
plt.xlabel('Distribution')
plt.ylabel('Unit Prices')
```

## 2. Bivariate Analysis

- Countplot (Categorical vs Categorical)

```python
size_order = ['S', 'M', 'L', 'XL', 'XXL']

sns.countplot(x='pizza_size', hue='pizza_category', data=df, order=size_order)
plt.xlabel('Pizza Size')
plt.ylabel('No of Pizza sold')
plt.title('Pizza Sizes vs Pizza Category')
```

- Barplot (Categorical vs Continuous)

```python
sns.barplot(x='pizza_name', y='total_price', data=df, estimator=sum, palette='Blues')
plt.title('Pizza Name vs Total Price')
plt.xticks(rotation=90)
plt.xlabel('Pizza Name')
plt.ylabel('Total Price (Revenue)')
```

- Line Plot (Categorical vs COntinuous)

```python
df.head(1)
```

```python
df['order_hour'] = pd.to_datetime(df['order_time']).dt.hour
```

```python
df.sample()
```

```python
# Peak hours for pizza order

sns.lineplot(x='order_hour', y='quantity', data=df, estimator='sum')
```

#### 3. Multivariate Analysis

```python
g = sns.FacetGrid(df, col='pizza_size', hue='pizza_category', col_wrap=1)  # No of graphs = pizza_size, hue = pizza_category
g.map(sns.lineplot, 'order_hour', 'unit_price')  # x-axis: order_hour  , y-axis: unit_price
g.add_legend()
```

- pairplot (FOr numerical columns)

```python
sns.pairplot(df, hue='pizza_size')
```

- Heatmap (Numerical Columns)

```python
corr = df[['quantity', 'total_price', 'unit_price']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
```
